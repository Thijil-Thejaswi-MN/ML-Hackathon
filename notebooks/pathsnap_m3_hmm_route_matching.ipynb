```python

from pathlib import Path
import shutil

PROJECT = Path(
    "/content/drive/MyDrive/Colab Notebooks/ML_hackathon/pathsnap"
)

graph_file = PROJECT / "src/pathsnap/graph.py"

# Keep a backup before changing anything
backup_file = graph_file.with_name("graph.py.backup")
shutil.copy2(graph_file, backup_file)

code = '''"""Road-network structures. OWNER: M3.

Derived only from trip GPS/network inputs. No ground-truth route access.
"""
from __future__ import annotations

from dataclasses import dataclass, field
from typing import Any

import numpy as np
from scipy.sparse import csr_matrix
from scipy.sparse.csgraph import dijkstra


@dataclass
class RoadGraph:
    """Road-network representation for one trip."""

    xy_nodes: np.ndarray
    arcs: np.ndarray
    arc_len: np.ndarray
    arc_bearing: np.ndarray

    # Directed node graph: adjacency[u, v] = minimum arc length u -> v
    adjacency: csr_matrix

    # reverse_twin[a] is the arc travelling in the opposite direction,
    # or -1 when no reverse arc exists.
    reverse_twin: np.ndarray

    # Outgoing and incoming arc IDs for each node.
    node_to_arcs: list[np.ndarray]
    node_to_in_arcs: list[np.ndarray]

    # Cache: (source tuple, cutoff) -> (distances, predecessors)
    _shortest_cache: dict = field(default_factory=dict, repr=False)


def build_graph(trip, projector) -> RoadGraph:
    """Build a directed, length-weighted graph from TripData."""
    nodes_lonlat = np.asarray(trip.nodes, dtype=float)
    arcs = np.asarray(trip.arcs, dtype=np.int64)

    if nodes_lonlat.ndim != 2 or nodes_lonlat.shape[1] != 2:
        raise ValueError("trip.nodes must have shape (n_nodes, 2)")
    if arcs.ndim != 2 or arcs.shape[1] != 2:
        raise ValueError("trip.arcs must have shape (n_arcs, 2)")
    if arcs.size and (arcs.min() < 0 or arcs.max() >= len(nodes_lonlat)):
        raise ValueError("Arc contains an invalid node ID")

    xy_nodes = np.asarray(
        projector.forward(nodes_lonlat[:, 0], nodes_lonlat[:, 1]),
        dtype=float,
    )

    if xy_nodes.shape != (len(nodes_lonlat), 2):
        raise ValueError("projector.forward must return shape (n_nodes, 2)")

    if len(arcs):
        vectors = xy_nodes[arcs[:, 1]] - xy_nodes[arcs[:, 0]]
        arc_len = np.hypot(vectors[:, 0], vectors[:, 1])
        arc_bearing = np.arctan2(vectors[:, 1], vectors[:, 0])
    else:
        arc_len = np.empty(0, dtype=float)
        arc_bearing = np.empty(0, dtype=float)

    n_nodes = len(nodes_lonlat)
    n_arcs = len(arcs)

    # Build outgoing/incoming arc lookup tables.
    outgoing = [[] for _ in range(n_nodes)]
    incoming = [[] for _ in range(n_nodes)]

    # Keep the first reverse arc found for each directed node pair.
    pair_to_arc = {}
    for arc_id, (u, v) in enumerate(arcs):
        outgoing[int(u)].append(arc_id)
        incoming[int(v)].append(arc_id)
        pair_to_arc.setdefault((int(u), int(v)), arc_id)

    reverse_twin = np.full(n_arcs, -1, dtype=np.int64)
    for arc_id, (u, v) in enumerate(arcs):
        reverse_twin[arc_id] = pair_to_arc.get((int(v), int(u)), -1)

    node_to_arcs = [
        np.asarray(ids, dtype=np.int64) for ids in outgoing
    ]
    node_to_in_arcs = [
        np.asarray(ids, dtype=np.int64) for ids in incoming
    ]

    # CSR adjacency for efficient shortest-path queries.
    adjacency = csr_matrix(
        (arc_len, (arcs[:, 0], arcs[:, 1])),
        shape=(n_nodes, n_nodes),
        dtype=float,
    )
    adjacency.sum_duplicates()
    adjacency.sort_indices()

    return RoadGraph(
        xy_nodes=xy_nodes,
        arcs=arcs,
        arc_len=arc_len,
        arc_bearing=arc_bearing,
        adjacency=adjacency,
        reverse_twin=reverse_twin,
        node_to_arcs=node_to_arcs,
        node_to_in_arcs=node_to_in_arcs,
    )


def bounded_shortest_paths(
    graph: RoadGraph,
    sources,
    cutoff_m: float,
):
    """Return (distances, predecessors) from source nodes.

    Results have shape (n_sources, n_nodes). Distances beyond cutoff_m
    are infinity. Predecessors are -9999 where no predecessor exists.

    Results are memoised per graph, source set and distance cutoff.
    """
    cutoff_m = float(cutoff_m)
    if cutoff_m < 0 or not np.isfinite(cutoff_m):
        raise ValueError("cutoff_m must be a finite non-negative number")

    if np.isscalar(sources):
        source_ids = (int(sources),)
    else:
        source_ids = tuple(int(s) for s in sources)

    if not source_ids:
        raise ValueError("At least one source node is required")

    n_nodes = graph.adjacency.shape[0]
    if any(s < 0 or s >= n_nodes for s in source_ids):
        raise ValueError("Source node ID is outside the graph")

    cache_key = (source_ids, cutoff_m)
    if cache_key in graph._shortest_cache:
        distances, predecessors = graph._shortest_cache[cache_key]
        return distances.copy(), predecessors.copy()

    distances, predecessors = dijkstra(
        graph.adjacency,
        directed=True,
        indices=np.asarray(source_ids, dtype=np.int64),
        return_predecessors=True,
        limit=cutoff_m,
    )

    # Keep a consistent 2-D result, even for one source.
    distances = np.atleast_2d(distances)
    predecessors = np.atleast_2d(predecessors)

    graph._shortest_cache[cache_key] = (
        distances.copy(),
        predecessors.copy(),
    )
    return distances, predecessors
'''

graph_file.write_text(code, encoding="utf-8")

print("Updated:", graph_file)
print("Backup:", backup_file)
print("Lines:", len(code.splitlines()))
print("\nFirst 30 lines:")
print("\n".join(code.splitlines()[:30]))
```

### [Step 10] Handling Duplicate Road Graph Edges
Applying a patch to resolve duplicate-edge handling inside the road graph build function.

```python
path = PROJECT / "src/pathsnap/graph.py"
text = path.read_text()

old = '''    adjacency = csr_matrix(
        (arc_len, (arcs[:, 0], arcs[:, 1])),
        shape=(n_nodes, n_nodes),
        dtype=float,
    )
    adjacency.sum_duplicates()
    adjacency.sort_indices()
'''

new = '''    edge_weights = {}
    for (u, v), length in zip(arcs, arc_len):
        key = (int(u), int(v))
        edge_weights[key] = min(edge_weights.get(key, float("inf")), float(length))

    if edge_weights:
        rows = [key[0] for key in edge_weights]
        cols = [key[1] for key in edge_weights]
        values = list(edge_weights.values())
        adjacency = csr_matrix(
            (values, (rows, cols)),
            shape=(n_nodes, n_nodes),
            dtype=float,
        )
    else:
        adjacency = csr_matrix((n_nodes, n_nodes), dtype=float)

    adjacency.sort_indices()
'''

assert old in text, "Could not find the expected adjacency block"
path.write_text(text.replace(old, new))
print("Duplicate-edge handling patched.")
```

### [Step 9] Road Graph Structure & Shortest Path Implementation

**Purpose:**
Build a robust directed, length-weighted road network graph representation and implement memoised, bounded shortest-path searching via SciPy.

**What this code does:**
- Replaces `src/pathsnap/graph.py` with an implementation of `RoadGraph` and `build_graph`.
- Integrates Dijkstra's algorithm from `scipy.sparse.csgraph` with distance boundaries (`cutoff_m`) to limit computation.
- Memoises shortest-path lookups inside `_shortest_cache` to speed up HMM transition calculations dramatically.

### [Step 9] Road Graph Structure & Shortest Path Implementation
Updating `src/pathsnap/graph.py` with directed road network utilities and Scipy-based dijkstra search.

```python

from pathlib import Path
import shutil
import textwrap

PROJECT = Path("/content/drive/MyDrive/Colab Notebooks/ML_hackathon/pathsnap")
target = PROJECT / "src/pathsnap/decoding/hmm.py"

if target.exists():
    backup = target.with_suffix(".py.backup")
    shutil.copy2(target, backup)
    print("Backup created:", backup)

code = r'''
"""HMM + Viterbi decoder. OWNER: M3.

States are candidate arcs for each observation.
Emissions use candidate probabilities, optionally weighted by GPS distance.
Transitions compare network distance with observed GPS displacement.
"""

from __future__ import annotations

from dataclasses import dataclass, field
from time import perf_counter

import numpy as np

from ..graph import bounded_shortest_paths


@dataclass
class DecodeResult:
    arc_per_obs: np.ndarray
    matched: np.ndarray
    segments: list = field(default_factory=list)
    route_arcs: list = field(default_factory=list)
    route_polyline: list = field(default_factory=list)
    breaks: list = field(default_factory=list)
    timings: dict = field(default_factory=dict)


def _project_points(track, projector):
    """Project track lon/lat coordinates into local metres."""
    xy = np.asarray(projector.forward(track[:, 0], track[:, 1]), dtype=float)
    if xy.shape != (len(track), 2):
        raise ValueError("projector.forward() must return an (n_obs, 2) array")
    return xy


def _arc_fractions(obs_xy, graph):
    """Projection fraction of every observation onto every directed arc."""
    arcs = np.asarray(graph.arcs, dtype=int)
    nodes = np.asarray(graph.xy_nodes, dtype=float)

    starts = nodes[arcs[:, 0]]
    ends = nodes[arcs[:, 1]]
    vectors = ends - starts
    denom = np.sum(vectors * vectors, axis=1)
    denom = np.maximum(denom, 1e-12)

    offsets = obs_xy[:, None, :] - starts[None, :, :]
    fractions = np.sum(offsets * vectors[None, :, :], axis=2) / denom[None, :]
    return np.clip(fractions, 0.0, 1.0)


def _network_distance(arc_a, frac_a, arc_b, frac_b, graph, cutoff_m):
    """Approximate along-network distance from a position on arc_a to arc_b."""
    arc_a = int(arc_a)
    arc_b = int(arc_b)
    lengths = np.asarray(graph.arc_len, dtype=float)
    arcs = np.asarray(graph.arcs, dtype=int)

    if arc_a == arc_b:
        return abs(float(frac_b) - float(frac_a)) * lengths[arc_a]

    end_node = int(arcs[arc_a, 1])
    start_node = int(arcs[arc_b, 0])

    if end_node == start_node:
        middle = 0.0
    else:
        distances, _ = bounded_shortest_paths(
            graph, [end_node], float(cutoff_m)
        )
        middle = float(distances[0, start_node])

        if not np.isfinite(middle):
            return np.inf

    return (
        (1.0 - float(frac_a)) * lengths[arc_a]
        + middle
        + float(frac_b) * lengths[arc_b]
    )


def _emission_log_probs(rows, obs_idx, fractions, graph, cfg):
    """Calculate normalized log emission probabilities for one observation."""
    probabilities = np.asarray(rows["p"], dtype=float)
    probabilities = np.clip(probabilities, 1e-12, None)

    log_emission = np.log(probabilities)

    weight = float(cfg.get("emission_distance_weight", 0.0))
    if weight:
        distances = np.asarray(rows["dist_m"], dtype=float)
        sigma = max(float(cfg.get("emission_sigma_m", 20.0)), 1e-9)
        log_emission -= weight * (distances ** 2) / (2.0 * sigma ** 2)

    # Normalize within this observation's candidate set.
    maximum = np.max(log_emission)
    log_emission -= maximum + np.log(np.exp(log_emission - maximum).sum())
    return log_emission


def _transition_log_score(
    arc_a, frac_a, arc_b, frac_b, gps_distance, delta_t, graph, cfg
):
    beta = max(float(cfg.get("beta_m", 50.0)), 1e-9)
    epsilon = float(cfg.get("epsilon", 1e-6))
    v_max = float(cfg.get("v_max_mps", 40.0))
    slack = float(cfg.get("dijkstra_slack_m", 100.0))

    cutoff = max(0.0, v_max * max(delta_t, 0.0) + slack)

    network_distance = _network_distance(
        arc_a, frac_a, arc_b, frac_b, graph, cutoff
    )

    if not np.isfinite(network_distance):
        return np.log(max(epsilon, 1e-300))

    score = -abs(network_distance - gps_distance) / beta

    # Penalize movement backwards along the same directed arc.
    if int(arc_a) == int(arc_b) and frac_b < frac_a:
        score -= float(cfg.get("backward_penalty", 5.0))

    return score


def _viterbi_run(obs_indices, rows_by_obs, obs_xy, fractions, track, graph, cfg):
    """Decode one uninterrupted run of observations with candidates."""
    states = []
    emissions = []

    for obs_idx in obs_indices:
        rows = rows_by_obs[obs_idx].sort_values("arc_id").drop_duplicates("arc_id")
        arc_ids = rows["arc_id"].to_numpy(dtype=int)
        states.append(arc_ids)
        emissions.append(
            _emission_log_probs(rows, obs_idx, fractions, graph, cfg)
        )

    deltas = [emissions[0].copy()]
    backpointers = [np.full(len(states[0]), -1, dtype=int)]

    for t in range(1, len(obs_indices)):
        previous_idx = obs_indices[t - 1]
        current_idx = obs_indices[t]

        gps_distance = float(np.linalg.norm(
            obs_xy[current_idx] - obs_xy[previous_idx]
        ))
        delta_t = float(track[current_idx, 2] - track[previous_idx, 2])

        current_delta = np.full(len(states[t]), -np.inf)
        current_back = np.full(len(states[t]), -1, dtype=int)

        for j, arc_b in enumerate(states[t]):
            scores = []
            for i, arc_a in enumerate(states[t - 1]):
                transition = _transition_log_score(
                    arc_a,
                    fractions[previous_idx, arc_a],
                    arc_b,
                    fractions[current_idx, arc_b],
                    gps_distance,
                    delta_t,
                    graph,
                    cfg,
                )
                scores.append(deltas[t - 1][i] + transition)

            scores = np.asarray(scores)
            best_previous = int(np.argmax(scores))
            current_delta[j] = scores[best_previous] + emissions[t][j]
            current_back[j] = best_previous

        deltas.append(current_delta)
        backpointers.append(current_back)

    # Backtrack from the most likely final state.
    path = np.empty(len(obs_indices), dtype=int)
    path[-1] = int(np.argmax(deltas[-1]))

    for t in range(len(obs_indices) - 1, 0, -1):
        path[t - 1] = backpointers[t][path[t]]

    return [int(states[t][path[t]]) for t in range(len(path))]


def decode(cands, trip, graph, projector, cfg: dict) -> DecodeResult:
    """Decode candidate arcs into one arc assignment per GPS observation."""
    started = perf_counter()

    n_obs = int(trip.n_obs)
    track = np.asarray(trip.track, dtype=float)

    if track.shape[0] != n_obs or track.shape[1] < 3:
        raise ValueError("trip.track must have shape (n_obs, >=3)")

    required = {"obs_idx", "arc_id", "p"}
    missing = required - set(cands.columns)
    if missing:
        raise ValueError(f"Candidate table is missing columns: {sorted(missing)}")

    if len(cands) and "dist_m" not in cands.columns:
        raise ValueError("Candidate table must include dist_m")

    if len(cands):
        if (cands["obs_idx"].min() < 0 or cands["obs_idx"].max() >= n_obs):
            raise ValueError("Candidate obs_idx is outside the trip")
        if (cands["arc_id"].min() < 0 or
                cands["arc_id"].max() >= len(graph.arcs)):
            raise ValueError("Candidate arc_id is outside the graph")

    obs_xy = _project_points(track, projector)
    fractions = _arc_fractions(obs_xy, graph)

    rows_by_obs = {
        int(idx): group
        for idx, group in cands.groupby("obs_idx", sort=True)
    }

    arc_per_obs = np.full(n_obs, -1, dtype=int)
    matched = np.zeros(n_obs, dtype=bool)

    # Build uninterrupted runs. Empty candidate observations and large time
    # gaps terminate a run and are reported as breaks.
    candidate_indices = sorted(rows_by_obs)
    runs = []
    current_run = []
    t_gap = float(cfg.get("t_gap_s", 30.0))

    for obs_idx in candidate_indices:
        if not current_run:
            current_run = [obs_idx]
            continue

        previous_idx = current_run[-1]
        delta_t = float(track[obs_idx, 2] - track[previous_idx, 2])

        if obs_idx != previous_idx + 1 or delta_t <= 0 or delta_t > t_gap:
            runs.append(current_run)
            current_run = [obs_idx]
        else:
            current_run.append(obs_idx)

    if current_run:
        runs.append(current_run)

    breaks = []
    segments = []

    for run_number, run in enumerate(runs):
        decoded = _viterbi_run(
            run, rows_by_obs, obs_xy, fractions, track, graph, cfg
        )
        arc_per_obs[np.asarray(run, dtype=int)] = decoded
        matched[np.asarray(run, dtype=int)] = True
        segments.append((int(run[0]), int(run[-1])))

        if run_number > 0:
            breaks.append(int(run[0]))

    # Fill unmatched observations with the nearest decoded observation's arc.
    # matched remains False, so downstream code can distinguish these guesses.
    valid = np.flatnonzero(matched)
    if len(valid):
        for obs_idx in np.flatnonzero(~matched):
            nearest = valid[np.argmin(np.abs(valid - obs_idx))]
            arc_per_obs[obs_idx] = arc_per_obs[nearest]

    # Preserve chronological decoded arc sequence; route stitching is handled
    # by route.py in the next implementation step.
    route_arcs = []
    for arc_id in arc_per_obs:
        if arc_id >= 0 and (not route_arcs or route_arcs[-1] != int(arc_id)):
            route_arcs.append(int(arc_id))

    route_polyline = []
    if route_arcs:
        arcs = np.asarray(graph.arcs, dtype=int)
        node_xy = np.asarray(graph.xy_nodes, dtype=float)
        node_ids = [int(arcs[route_arcs[0], 0])]
        node_ids.extend(int(arcs[a, 1]) for a in route_arcs)
        lonlat = np.asarray(projector.inverse(node_xy[node_ids]), dtype=float)
        route_polyline = [
            [float(lat), float(lon)] for lon, lat in lonlat
        ]

    return DecodeResult(
        arc_per_obs=arc_per_obs,
        matched=matched,
        segments=segments,
        route_arcs=route_arcs,
        route_polyline=route_polyline,
        breaks=breaks,
        timings={"decode_seconds": perf_counter() - started},
    )
'''

target.parent.mkdir(parents=True, exist_ok=True)
target.write_text(textwrap.dedent(code).lstrip(), encoding="utf-8")
print("Updated:", target)
print("Lines:", len(target.read_text().splitlines()))
```

```python

path = PROJECT / "src/pathsnap/decoding/hmm.py"
source = path.read_text()

source = source.replace(
'''def _arc_fractions(obs_xy, graph):
    """Projection fraction of every observation onto every directed arc."""
    arcs = np.asarray(graph.arcs, dtype=int)
    nodes = np.asarray(graph.xy_nodes, dtype=float)

    starts = nodes[arcs[:, 0]]
    ends = nodes[arcs[:, 1]]
    vectors = ends - starts
    denom = np.sum(vectors * vectors, axis=1)
    denom = np.maximum(denom, 1e-12)

    offsets = obs_xy[:, None, :] - starts[None, :, :]
    fractions = np.sum(offsets * vectors[None, :, :], axis=2) / denom[None, :]
    return np.clip(fractions, 0.0, 1.0)
''',
'''def _candidate_fractions(obs_xy, graph, cands):
    """Compute projection fractions only for arcs present in the candidate table."""
    arcs = np.asarray(graph.arcs, dtype=int)
    nodes = np.asarray(graph.xy_nodes, dtype=float)
    result = {}

    for obs_idx, group in cands.groupby("obs_idx"):
        point = obs_xy[int(obs_idx)]
        for arc_id in group["arc_id"].unique():
            arc_id = int(arc_id)
            start = nodes[arcs[arc_id, 0]]
            end = nodes[arcs[arc_id, 1]]
            vector = end - start
            denom = max(float(np.dot(vector, vector)), 1e-12)
            fraction = float(np.dot(point - start, vector) / denom)
            result[(int(obs_idx), arc_id)] = float(np.clip(fraction, 0.0, 1.0))

    return result
''')

source = source.replace(
    'fractions[current_idx, arc_b]',
    'fractions[(current_idx, int(arc_b))]'
)
source = source.replace(
    'fractions[previous_idx, arc_a]',
    'fractions[(previous_idx, int(arc_a))]'
)
source = source.replace(
    'fractions = _arc_fractions(obs_xy, graph)',
    'fractions = _candidate_fractions(obs_xy, graph, cands)'
)

path.write_text(source)
print("Applied candidate-only fraction calculation.")
```
